## Processing results

```python
arr = np.loadtxt("normal_decomp_hopf.txt")
w_range = arr[:,0]
normal_decomp_res = arr[:,1:]

arr = np.loadtxt("lr_decomp_hopf.txt")
lr_decomp_res = arr[:,1:]
```

```python
fig, ax = plt.subplots(1,2)
fig.set_size_inches(12,4)
for i in range(0,100):
    ax[0].scatter(w_range,normal_decomp_res[:,i], c = 'k', s= 1)
    ax[1].scatter(w_range,lr_decomp_res[:,i], c = 'k', s= 1)


ax[0].set_xlabel(r'$\omega$')
ax[1].set_xlabel(r'$\omega$')
ax[0].set_ylabel(r'$\frac{|W-C|}{|C|+|W-C|}$', fontsize = 15)
ax[1].set_ylabel(r'$\frac{|\Gamma\Pi|}{|\Gamma\Omega|+|\Gamma\Pi|}$', fontsize = 15)

sns.despine()
plt.tight_layout()
plt.show()
```

```python
arr = np.loadtxt("hhd_hopf_fast.txt")
w_range = arr[:,0]
hhd_decomp_res = arr[:,1:]

arr = np.loadtxt("hhd_rho_hopf_fast.txt")
hhd_rho_res = arr[:,1:]
```

```python
fig, ax = plt.subplots(1,2)
fig.set_size_inches(12,4)
for i in range(0,100):
    ax[0].scatter(w_range,hhd_decomp_res[:,i], c = 'k', s= 1)
    ax[1].scatter(w_range,hhd_rho_res[:,i], c = 'k', s= 1)


ax[0].set_xlabel(r'$\omega$')
ax[1].set_xlabel(r'$\omega$')
ax[0].set_ylabel(r'$\frac{|W_{\text{irr}}|}{|W_{\text{rev}}|+|W_{\text{irr}}|}$', fontsize = 15)
ax[1].set_ylabel(r'$\rho$', fontsize = 15)

sns.despine()
plt.tight_layout()
plt.show()
```
